```python
import os
import shutil
import random

source = "/content/deepfake_dataset/Final Dataset"
destination = "/content/deepfake_data"

random.seed(42)

for label in ["Fake", "Real"]:
    source_folder = os.path.join(source, label)

    images = [
        f for f in os.listdir(source_folder)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ]

    random.shuffle(images)

    total = len(images)
    train_end = int(total * 0.80)
    val_end = int(total * 0.90)

    splits = {
        "train": images[:train_end],
        "validation": images[train_end:val_end],
        "test": images[val_end:]
    }

    for split, files in splits.items():
        target_folder = os.path.join(destination, split, label)

        for filename in files:
            shutil.copy2(
                os.path.join(source_folder, filename),
                os.path.join(target_folder, filename)
            )

    print(f"{label}: {total} images split")

print("Dataset splitting completed!")
```

```python
import tensorflow as tf

IMG_SIZE = (224, 224)
BATCH_SIZE = 32

train_ds = tf.keras.utils.image_dataset_from_directory(
    "/content/deepfake_data/train",
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="binary",
    shuffle=True
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    "/content/deepfake_data/validation",
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="binary",
    shuffle=False
)

test_ds = tf.keras.utils.image_dataset_from_directory(
    "/content/deepfake_data/test",
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="binary",
    shuffle=False
)

print("Images loaded successfully!")
print("Classes:", train_ds.class_names)
```

```python
AUTOTUNE = tf.data.AUTOTUNE

train_ds = train_ds.prefetch(buffer_size=AUTOTUNE)
val_ds = val_ds.prefetch(buffer_size=AUTOTUNE)
test_ds = test_ds.prefetch(buffer_size=AUTOTUNE)

print("Dataset pipeline optimized!")
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models

base_model = tf.keras.applications.MobileNetV2(
    input_shape=(224, 224, 3),
    include_top=False,
    weights="imagenet"
)

# Freeze the pretrained layers
base_model.trainable = False

model = models.Sequential([
    layers.Input(shape=(224, 224, 3)),
    layers.Rescaling(1./255),
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.Dropout(0.3),
    layers.Dense(1, activation="sigmoid")
])

model.summary()
```

```python
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.0001),
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

print("Model compiled successfully!")
```

```python
history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=5
)
```

```python
test_loss, test_accuracy = model.evaluate(test_ds)

print(f"Test Loss: {test_loss:.4f}")
print(f"Test Accuracy: {test_accuracy * 100:.2f}%")
```

```python
import gradio as gr
import numpy as np
import cv2
from PIL import Image

def detect_image(image):
    if image is None:
        return "Please upload an image.", ""

    image = image.resize((224, 224))
    img_array = np.array(image).astype("float32")
    img_array = np.expand_dims(img_array, axis=0)

    prediction = model_90.predict(img_array, verbose=0)[0][0]

    if prediction >= 0.5:
        result = "REAL"
        score = prediction * 100
    else:
        result = "POTENTIALLY FAKE"
        score = (1 - prediction) * 100

    return result, f"{score:.2f}%"
```

```python
def detect_video(video_path):
    if video_path is None:
        return "Please upload a video.", ""

    cap = cv2.VideoCapture(video_path)

    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    fps = cap.get(cv2.CAP_PROP_FPS)

    # Analyze about 20 frames from the video
    frame_indices = np.linspace(
        0,
        max(total_frames - 1, 0),
        min(20, total_frames),
        dtype=int
    )

    predictions = []

    for frame_index in frame_indices:
        cap.set(cv2.CAP_PROP_POS_FRAMES, int(frame_index))

        success, frame = cap.read()

        if not success:
            continue

        # OpenCV BGR → RGB
        frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

        # Resize for MobileNetV2
        frame = cv2.resize(frame, (224, 224))

        frame = frame.astype("float32")
        frame = np.expand_dims(frame, axis=0)

        prediction = model_90.predict(frame, verbose=0)[0][0]

        predictions.append(prediction)

    cap.release()

    if len(predictions) == 0:
        return "Could not read video.", ""

    # Average prediction across analyzed frames
    average_prediction = np.mean(predictions)

    if average_prediction >= 0.5:
        result = "REAL"
        score = average_prediction * 100
    else:
        result = "POTENTIALLY FAKE"
        score = (1 - average_prediction) * 100

    return result, f"{score:.2f}%"
```

```python
import os
import shutil
import random

source = "/content/deepfake_dataset/Final Dataset"
destination = "/content/deepfake_data"

random.seed(42)

for label in ["Fake", "Real"]:
    source_folder = os.path.join(source, label)
    images = [
        f for f in os.listdir(source_folder)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ]

    random.shuffle(images)

    total = len(images)
    train_end = int(total * 0.80)
    val_end = int(total * 0.90)

    splits = {
        "train": images[:train_end],
        "validation": images[train_end:val_end],
        "test": images[val_end:]
    }

    for split, files in splits.items():
        target_folder = os.path.join(destination, split, label)

        for filename in files:
            shutil.copy2(
                os.path.join(source_folder, filename),
                os.path.join(target_folder, filename)
            )

    print(
        label,
        "→",
        len(splits["train"]), "train,",
        len(splits["validation"]), "validation,",
        len(splits["test"]), "test"
    )

print("Dataset split completed!")
```
